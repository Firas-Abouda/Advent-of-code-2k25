```python
import re


ids = []
ranges = []

with open('input.txt', 'r') as file:
    for line in file:
        if line != '\n':
            res = list(map(int,re.split(r'[-,\n]', line)[:-1]))
            if len(res) ==2 :
                ranges.append(res)
            elif len(res)  :
                ids.append(res[0])
```

```python
ranges.sort(key=lambda x: x[0]) 
merged_ranges = []

merged_ranges.append(ranges[0])
for range in ranges:
    if range[0] <= merged_ranges[-1][1]:
        if range[1] > merged_ranges[-1][1]:
            merged_ranges[-1][1] = range[1]
    else :
        merged_ranges.append(range)
```

```python
# a very naive iteration on id and (marged) range, having a tab of all ids is not faisable because

def aoc1():
    count = 0
    for id in ids:
        for inf, sup in merged_ranges:
            if inf <= id and sup >= id:
                count+=1
                break

    return count

aoc1() #789
```

```python
# well part 2 has beeen done in processing data already

def aoc2():
    sum = 0 
    for inf, sup in merged_ranges:
        sum += (sup -inf) + 1
    return sum

aoc2() #343329651880509
```
