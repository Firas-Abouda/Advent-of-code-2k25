```python
L = []; ope = []
i = 1
with open('input.txt', 'r') as file:
    lines = file.readlines()
for line in lines[:-1]:
    L.append(list(map(float, line.split())))

ope = lines[-1].split()
```

```python
# ord('*')= 42 , ord('+') = 43
import numpy as np 

ope_array = np.array(ope)
nb_array = np.array(L)

sort_idx = np.argsort(ope_array)
nb_array_sorted = nb_array[:, sort_idx] #we sort and order operation so we can use vectorize operation and make code simppler to understand
```

```python
def aoc1():
    slice_idx = list(ope_array[sort_idx]).index('+')
    mul_vect = nb_array_sorted[0,:slice_idx]; add_vect = nb_array_sorted[0,slice_idx:]
    for i in range (1,nb_array_sorted.shape[0]):
        mul_vect = mul_vect * nb_array_sorted[i,:slice_idx] ; add_vect = add_vect + nb_array_sorted[i,slice_idx:]

    return int(mul_vect.sum() + add_vect.sum())

aoc1()
```

```python
# let's keep this part2 simple and just rea the file and iteratively make the operations 

def aoc2():
    total = 0.0
    current = 0.0
    buffer = ''
    sign = ''
    for k in range(len(lines[0])-1):
        for i in range(0,len(lines)-1):
            buffer += lines[i][k]
        
        if buffer ==' '*(len(lines)-1):
            total += current
            sign = ''
            current = 0.0
        else : 
            match(sign):
                case '':
                    current = float(buffer)
                case '+':
                    current += float(buffer)
                case '*':
                    current *= float(buffer)
        buffer = ''
        if lines[-1][k] != ' ':
            sign  =  lines[-1][k]

    return int(total)

aoc2()
```
