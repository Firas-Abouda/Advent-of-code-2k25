```python
banks = []

with open('input.txt', 'r') as file:
    for line in file:
        banks.append(list(map(int, line[:-1])))
        
```

```python
# luckely in python np.argmax return first occurences if multiple occurence therefore we get part1 pretty directly
import numpy as np 
def aoc1():
    sum = 0
    for bank in banks:
        ten_idx = np.argmax(bank[:-1])
        digit_idx = np.argmax(bank[ten_idx+1:]) + ten_idx +1
        sum += bank[ten_idx] *10 + bank[digit_idx]
    return sum

aoc1()
```

```python
def aoc2():
    sum = 0
    for bank in banks:
        sum_idx = 0
        for i in range(11,-1,-1):
            if i > 0 :
                max_idx = np.argmax(bank[sum_idx :-i])
            else : 
                max_idx = np.argmax(bank[sum_idx :])
            sum_idx += max_idx 
            sum += bank[sum_idx] * (10**i)
            sum_idx +=1
    return sum

aoc2()
```
