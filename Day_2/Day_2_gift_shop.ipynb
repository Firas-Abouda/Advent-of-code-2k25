```python
def get_range_int(s):
    l = s.split('-')
    begin = int(l[0]); end = int(l[1])
    return begin, end
```

```python
L= []
for r in ranges:
    _,e = get_range_int(r)
    L.append(e)
L.sort()
L[-1]
```

```python
invalid_ids = []
for i in range(1,100000):
    s = str(i)+str(i)
    invalid_ids.append(int(s))




def gen_invalid_id(begin, end,supp ): 
    sum = 0 
    invalid_ids = []
    if begin == end: 
        s =str(end)+str(end)
        if int(s) < supp:
            invalid_ids.append(int(s))
            sum = int(s)
    else : 
        for i in range (begin,end): 
            s = str(i)+str(i)
            invalid_ids.append(int(s))
            sum += int(s)

    if supp%2== 0 and str(supp)[-(len(str(supp))//2):] == str(supp)[:(len(str(supp))//2)]:
        invalid_ids.append(supp)
        sum+= supp
    return invalid_ids, sum
```
