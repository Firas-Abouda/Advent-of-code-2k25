```python
instructions = []

with open('input.txt', 'r') as file:
    for line in file:
        instructions.append(line[:-1])
        
```

```python
def read_instruction(s):
    direction = s[0]
    distance = int(s[1:])
    return direction, distance 

DIAL_SIZE = 100

def make_rotation(current_dial, direction, distance):
    match direction:
        case "L": 
            new_dial = (current_dial - distance) % DIAL_SIZE
        case "R" :
            new_dial = (current_dial + distance) % DIAL_SIZE
    return new_dial

```

```python
DIAL_START = 50 

def aoc1():
    dial = DIAL_START
    count = 0
    for instruction in instructions:
        direction, distance  = read_instruction(instruction)
        dial = make_rotation(dial, direction, distance)
        if dial ==0 : 
            count +=1 
    return count
```

```python
print(aoc1())
```

```python
def aoc2():
    dial = DIAL_START
    count = 0
    for instruction in instructions:
        direction, distance  = read_instruction(instruction)
        count += distance // DIAL_SIZE
        new_dial = make_rotation(dial, direction, distance)
        if (dial != 0 and (new_dial > dial and direction == 'L' or new_dial < dial and direction == 'R')) or new_dial ==0 : 
            count +=1 
        dial = new_dial 
    return count
```

```python
print(aoc2())
```
