```python
map=[]

with open('example.txt', 'r') as file:
    for line in file:
        map.append(list(line.split()[0]))

num_map  = [[0 if x == '.' else 1 for x in m] for m in map]
```

```python
# let's bring a tank to fight a mosquito xD
import torch.nn as nn 
import torch

tensor_map = torch.Tensor(num_map)

def aoc1():
    pad = nn.AvgPool2d(kernel_size = 3, stride = 1, padding = 1,divisor_override=1)
    tensor_map_4d = tensor_map.unsqueeze(0).unsqueeze(0) 
    sum_padded_map = pad(tensor_map_4d).squeeze(0).squeeze(0)
    forklift_adj = torch.where(tensor_map ==0, tensor_map, sum_padded_map )
    count = ((forklift_adj  > 0) & (forklift_adj  < 5)).sum()

    return count.item()

aoc1()
```

```python
# part 2 is just iterating part 1 

def aoc2():
    pad = nn.AvgPool2d(kernel_size = 3, stride = 1, padding = 1,divisor_override=1)
    map = tensor_map.copy()
    zeros = torch.zeros(tensor_map.shape)
    count = 1 
    sum = 0
    while count > 0 :
        tensor_map_4d = map.unsqueeze(0).unsqueeze(0) 
        sum_padded_map = pad(tensor_map_4d).squeeze(0).squeeze(0)
        forklift_adj = torch.where(tensor_map ==0, tensor_map, sum_padded_map )
        count = ((forklift_adj  > 0) & (forklift_adj  < 5)).sum()
        sum += count
        map = torch.where(forklift_adj >4, tensor_map, zeros)
```
